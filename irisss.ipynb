```python
import sklearn
import sklearn.datasets
from rich.console import Console
import pandas
import sklearn.model_selection
from xgboost import XGBClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report
from sklearn.linear_model import LinearRegression
from sklearn.naive_bayes import GaussianNB
import seaborn
import numpy as np
```

```python
console = Console()
# iris_ds = sklearn.datasets.load_iris()
iris_ds = sklearn.datasets.load_digits()
# console.print(iris_ds.keys())
# console.print(iris_ds)
# console.print(iris_ds.DESCR)
```

```python
df = pandas.DataFrame(iris_ds.data, columns=iris_ds.feature_names)
console.print(df.head())
console.print(df.info())
```

```python
X = iris_ds.data
y = iris_ds.target

X_train, X_test, y_train, y_test = sklearn.model_selection.train_test_split(X,y,test_size=0.1, random_state=42)
```

```python
# model = KNeighborsClassifier(n_neighbors=6)
# model = XGBClassifier(n_estimators=1000)
model = GaussianNB()

model.fit(X_train,y_train)
y_pred = model.predict(X_test)
# console.print(classification_report(y_test, y_pred, target_names=iris_ds.target_names))
console.print(classification_report(y_test, y_pred, target_names=[ str(x) for x in iris_ds.target_names]  ))
```

```python
model = KNeighborsClassifier(n_neighbors=2)
# model = XGBClassifier(n_estimators=1000)
# model = LinearRegression()

model.fit(X_train,y_train)
y_pred = model.predict(X_test)
console.print(classification_report(y_test, y_pred, target_names=[ str(x) for x in iris_ds.target_names]))
```

```python
# model = KNeighborsClassifier(n_neighbors=6)
model = XGBClassifier(n_estimators=1000)
# model = LinearRegression()

model.fit(X_train,y_train)
y_pred = model.predict(X_test)
console.print(classification_report(y_test, y_pred, target_names=[ str(x) for x in iris_ds.target_names]))
```
